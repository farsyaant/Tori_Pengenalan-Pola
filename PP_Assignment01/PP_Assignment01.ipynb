###*Text Data Preprocessing* dan Ekstraksi Ciri TF-IDF


####*Text Preprocessing*
***Text Preprocessing*** adalah tahapan yang mentransformasikan data yang tidak terstruktur (teks) menjadi terstruktur *(term index)*. *Term index* akan menjadi representasi data teks tersebut untuk dapat digunakan pada proses *mining* selanjutnya. Tahapan dalam *text preprocessing* adalah sebagai berikut.

#####**1. Parsing**: Penguraian struktur dokumen menjadi unit-unit terpisah

```python
def parsing(corpus_dict):
    doc_names = []
    documents = []
    sentences = {}

    for name, kalimat_list in corpus_dict.items():
        doc_names.append(name)
        documents.append(" ".join(kalimat_list))
        sentences[name] = kalimat_list

    return doc_names, documents, sentences
doc_names, documents, sentences = parsing(corpus_raw)
print("TAHAP 1: PARSING")
print ()

for i, (name, doc) in enumerate(zip(doc_names, documents)):
    kalimat_list = sentences[name]
    print(f"Artikel {i+1}: {name}")
    print(f"  Jumlah kalimat : {len(kalimat_list)}")
    print(f"  Jumlah kata    : {len(doc.split())}")
    print()
    for j, kal in enumerate(kalimat_list, 1):
        words = kal.split()
        line, lines = "", []
        for w in words:
            if len(line) + len(w) + 1 > 58:
                lines.append(line)
                line = w
            else:
                line = line + " " + w if line else w
        if line:
            lines.append(line)
        print(f"  K{j}: {lines[0]}")
        for l in lines[1:]:
            print(f"       {l}")
    print()
```

#####**2. Lexical Analysis/Tokenization**: Pemisahan kata per kata, penghilangan dan pembersihan karakter selain alfabet yang dianggap sebagai delimiter dan tidak ada hubungannya dengan informasi dokumen, serta *case folding.* <br>
*   Token: Pemisahan kata dari teks dengan mengabaikan duplikasi.
*   Type: Token yang mempertimbangkan duplikasi.
*   Term: Type setelah dilakukan stemming. <br>

```python
def lexical_analysis(text):
    # 1. Case Folding
    lowered_text = text.lower()

    # 2. Cleaning (Manual)
    cleaned_text = ""
    for char in lowered_text:
        if 'a' <= char <= 'z':
            cleaned_text += char
        else:
            cleaned_text += ' '

    # 3. Tokenization: List seluruh kata (Termasuk duplikasi)
    # Sesuai revisimu: Token = list seluruh kata
    token_list = cleaned_text.split()

    # 4. Type: List kata unik
    type_list = []
    for word in token_list:
        if word not in type_list:
            type_list.append(word)

    return token_list, type_list

print("TAHAP 2: LEXICAL ANALYSIS / TOKENIZATION")
print()

# Kita simpan hasil token (seluruh kata) untuk diproses di tahap selanjutnya
all_tokens_docs = []

for i, name in enumerate(doc_names):
    text_content = documents[i]
    tokens, types = lexical_analysis(text_content)
    all_tokens_docs.append(tokens)

    print(f"Artikel {i+1}: {name}")

    if i == 0:
        print(f"  Hasil Token:\n    {tokens}")
        print(f"\n  Hasil Type:\n    {types}")
        print(f"\n  Total Token: {len(tokens)}")
        print(f"  Total Type : {len(types)}")
    else:
        print(f"  Total Token: {len(tokens)}")
        print(f"  Total Type : {len(types)}")
    print()
```

#####**4. Phrase Detection:** Deteksi dua kata atau lebih yang menjadi frasa.
*   Metode Thesauri: Membandingkan frasa-frasa dalam suatu daftar dengan teks.
*   Metode Rule/Aturan: Dua kata yang sering muncul berurutan dianggap sebagai frasa.
*   Syntatic Analysis <br>

Nilai PMI (Pointwise Mutual Information) dan frekuensi minimum ditentukan di awal untuk menyaring *noise* karena corpus yang kecil dan dapat mengakibatkan nilai PMI dapat menjadi tidak stabil pada kata yang jarang muncul. Selain itu, nilai PMI 4.0 yang diambil berarti pasangan kata tersebut muncul 16 kali lebih sering daripada muncul secara terpisah.

```python
PMI_THRESHOLD = 4.0
MIN_FREQ      = 2

import math

def compute_pmi(filtered_docs, threshold=PMI_THRESHOLD, min_freq=MIN_FREQ):
    unigram_count = {}
    bigram_count  = {}
    total_uni = 0
    total_bi  = 0

    # Menghitung frekuensi unigram dan bigram secara manual
    for tokens in filtered_docs:
        for t in tokens:
            unigram_count[t] = unigram_count.get(t, 0) + 1
            total_uni += 1

        for j in range(len(tokens) - 1):
            pair = (tokens[j], tokens[j+1])
            bigram_count[pair] = bigram_count.get(pair, 0) + 1
            total_bi += 1

    if total_uni == 0 or total_bi == 0:
        return set()

    significant = set()
    for (a, b), cnt in bigram_count.items():
        if cnt < min_freq:
            continue

        p_ab = cnt / total_bi
        p_a  = unigram_count[a] / total_uni
        p_b  = unigram_count[b] / total_uni

        if p_a > 0 and p_b > 0:
            pmi = math.log2(p_ab / (p_a * p_b))
            if pmi >= threshold:
                significant.add((a, b))

    return significant

def phrase_detection(filtered_docs, significant_bigrams):
    phrased_docs = []
    for tokens in filtered_docs:
        result = []
        i = 0
        while i < len(tokens):
            if (i < len(tokens) - 1 and (tokens[i], tokens[i+1]) in significant_bigrams):
                result.append(tokens[i] + "_" + tokens[i+1])
                i += 2
            else:
                result.append(tokens[i])
                i += 1
        phrased_docs.append(result)
    return phrased_docs

significant_bigrams = compute_pmi(all_filtered_docs)
all_phrased_docs = phrase_detection(all_filtered_docs, significant_bigrams)

print(f"TAHAP 4: PHRASE DETECTION (PMI >= {PMI_THRESHOLD}, freq >= {MIN_FREQ})")
print()

if significant_bigrams:
    print(f"  Bigram signifikan terdeteksi ({len(significant_bigrams)}):")
    for bg in sorted(list(significant_bigrams)):
        print(f"    {bg[0]} + {bg[1]}  ->  {bg[0]}_{bg[1]}")
else:
    print("  Tidak ada bigram yang memenuhi syarat PMI + frekuensi.")

print()

for i, (name, after) in enumerate(zip(doc_names, all_phrased_docs)):
    phrases = [t for t in after if '_' in t]
    print(f"Artikel {i+1}: {name} -> {len(phrases)} frasa")
    if phrases:
        print(f"     {phrases}")
    print()
```

#####**5. Stemming dan Lemmatization:** Transformasi bentuk kata menjadi kata dasar dari hasil filtering.
*   Pendekatan Algoritmic: Mendeteksi imbuhan pada kata untuk dibuang. Berikut digunakan  algoritma Nazief-Adriani untuk membuang imbuhan (prefiks dan sufiks) secara mekanis. Sufiks, seperti "-lah", "-kah", "-i", dihapus terlebih dahulu dan dilanjutkan dengan penghapusan prefiks.
*   Lemmatization: Setelah stemming dilakukan, hasil stem dipetakan ke kata dasar yang valid di kamus, misal "kepunah" ke "punah", "ilik" ke "milik", dan lainnya.










```python
STEM_PROTECT = {
    "hutan", "kawasan", "konservasi", "habitat", "populasi",
    "spesies", "ancaman", "kepunahan", "degradasi", "konversi",
    "reproduksi", "infrastruktur", "satwa", "mamalia",
    "keanekaragaman", "hayati", "karbon", "koridor", "individu",
    "orangutan", "borneo", "sumatera", "kalimantan", "jawa",
    "iucn", "wwf", "dugong", "duyung", "gajah", "macan",
    "tutul", "leopard", "survei", "predator", "seleksi",
    "konsesi", "fragmentasi", "konektivitas", "komprehensif",
    "nasional", "stasiun", "resmi", "peran", "merah"
}

PREFIXES = [
    ("memper", ""), ("diper", ""), ("menge", ""), ("meny", "s"),
    ("meng", ""), ("men", ""), ("mem", ""), ("me", ""),
    ("peng", ""), ("peny", "s"), ("pem", ""), ("pen", ""),
    ("per", ""), ("ber", ""), ("ter", ""), ("di", ""),
    ("ke", ""), ("se", ""), ("pe", "")
]

def strip_suffix(word):
    for suf in ["kan", "an", "i"]:
        if word.endswith(suf) and len(word) - len(suf) >= 4:
            return word[:-len(suf)]
    return word

def strip_prefix(word):
    for pre, repl in PREFIXES:
        if word.startswith(pre) and len(word) - len(pre) + len(repl) >= 3:
            return repl + word[len(pre):]
    return word

def stem(word):
    if word in STEM_PROTECT or "_" in word:
        return word

    w = word
    if w.endswith("kan") and len(w) > 6: w = w[:-3]
    elif w.endswith("an") and len(w) > 5: w = w[:-2]
    elif w.endswith("i") and len(w) > 5 and not w.endswith("si"): w = w[:-1] # Jangan potong -si

    # Prefix stripping
    for pre, repl in PREFIXES:
        if w.startswith(pre) and len(w) - len(pre) + len(repl) >= 3:
            w = repl + w[len(pre):]
            break

    return w if len(w) >= 3 else word

print("TAHAP 5: STEMMING (Nazief-Adriani)")
print()

all_stemmed_docs = []
for i, name in enumerate(doc_names):
    current_docs = all_phrased_docs[i]
    stemmed = [stem(t) for t in current_docs]
    all_stemmed_docs.append(stemmed)

    changes = [(b, a) for b, a in zip(current_docs, stemmed) if b != a]

    print(f"Artikel {i+1}: {name}")
    if changes:
        for orig, st in changes[:5]:
            print(f"  {orig:<15} -> {st}")
    else:
        print("  (Tidak ada perubahan)")
    print()

LEMMA_DICT = {
    "sebab": "sebab",
    "akhir": "akhir",
    "impa": "timpa",
    "puluh": "puluh",
    "ratus": "ratus",
    "tumbuh": "tumbuh",
    "leks": "seleksi",
    "cekung": "cekung",
    "nelay": "nelayan",
    "angkap": "tangkap",
    "berada": "ada",
    "ilik": "milik",
    "henti": "henti",
    "jad": "jadi",
    "susut": "susut",
    "banyak": "banyak",
    "resm": "resmi",
    "laku": "laku",
    "inta": "intai",
    "amat": "amat"
}

print("TAHAP 6: LEMMATIZATION")
print()

all_lemmatized_docs = []
for i, name in enumerate(doc_names):
    current_stemmed = all_stemmed_docs[i]
    lemmatized = [LEMMA_DICT.get(t, t) for t in current_stemmed]
    all_lemmatized_docs.append(lemmatized)

    changes = [(s, l) for s, l in zip(current_stemmed, lemmatized) if s != l]

    print(f"Artikel {i+1}: {name}")
    if changes:
        for st, lem in changes[:5]:
            print(f"  {st:<15} -> {lem}")
    else:
        print("  (Tidak ada perubahan)")
    print()
```

### Ekstraksi Ciri TF-IDF

TF-IDF (Term Frequency dan Inverse Document Frequency) adalah salah satu teknik representasi data teks.
*   TF: Menunjukkan relevansi term pada suatu dokumen dengan kalkulasi *word count* atau frekuensi.
*   IDF: Mempertimbangkan kata yang tidak terlalu sering muncul, tetapi kemungkinan nilainya signifikan/relevan terhadap dokumen. <br>
-> Semakin jarang sebuah kata, semakin besar nilai IDF-nya dan sebaliknya.
*   TF-IDF: Mengalikan nilai TF dan IDF yang akan direpresentasikan sebagai vektor suatu dokumen.



```python
def compute_tf(tokens):
    total = len(tokens)
    counts = {}
    for t in tokens:
        counts[t] = counts.get(t, 0) + 1
    return {t: c / total for t, c in counts.items()}

def compute_idf(docs):
    N = len(docs)
    df_counts = {}

    all_terms = set()
    for doc in docs:
        for t in doc:
            all_terms.add(t)

    for t in all_terms:
        count = 0
        for doc in docs:
            if t in doc:
                count += 1
        df_counts[t] = count

    return {t: 1 + math.log(N / df_counts[t]) for t in df_counts}

def compute_tfidf(docs):
    idf_scores = compute_idf(docs)
    tfidf_matrix = []

    for tokens in docs:
        tf = compute_tf(tokens)
        doc_tfidf = {t: tf[t] * idf_scores[t] for t in tf}
        tfidf_matrix.append(doc_tfidf)

    return tfidf_matrix, idf_scores, sorted(idf_scores.keys())

tfidf_matrix, idf_scores, vocabulary = compute_tfidf(all_lemmatized_docs)

print("TAHAP 7: EKSTRAKSI CIRI — TF-IDF")
print()
print(f"  Ukuran vocabulary : {len(vocabulary)} term")
print(f"  Jumlah dokumen    : {len(all_lemmatized_docs)}")
print()

for i, (name, tfidf) in enumerate(zip(doc_names, tfidf_matrix)):
    top5 = sorted(tfidf.items(), key=lambda x: x[1], reverse=True)[:5]
    print(f"D{i+1} [{name}]  — top-5 term:")
    for rank, (term, score) in enumerate(top5, 1):
        print(f"     {rank}. {term:<30}  TF-IDF = {score:.4f}")
    print()

# Tabel term yang muncul di minimal 2 dokumen
multi_doc_terms = [t for t in vocabulary
                   if sum(1 for td in tfidf_matrix if t in td) >= 2]

print("-" * 65)
print("Tabel TF-IDF  (term muncul di >= 2 dokumen)")
print("-" * 65)
headers = "".join(f" {'D'+str(i+1):>7}" for i in range(len(doc_names)))
print(f"{'Term':<22}" + headers)
print("-" * 65)

for term in sorted(multi_doc_terms):
    row = f"""{term:<22}"""
    for tfidf in tfidf_matrix:
        row += f" {tfidf.get(term, 0.0):>7.4f}"
    print(row)
print()
```

**Cosine Similarity** adalah metode untuk mengukur tingkat similaritas atau kemiripian antara dua dokumen dengan menghitung nilai kosinus sudut di antara vektor-vektor TF-IDF dalam ruang multidimensi. Kesamaan tema atau kata kunci yang kuat ditunjukkan oleh skor yang mendekati angka 1, sementara tidak adanya keterkaitan informasi ditandai oleh skor yang mendekati 0.

```python
def calculate_cosine(vec_a, vec_b):
    all_terms = set(vec_a.keys()) | set(vec_b.keys())

    dot_product = sum(vec_a.get(t, 0) * vec_b.get(t, 0) for t in all_terms)

    norm_a = math.sqrt(sum(v**2 for v in vec_a.values()))
    norm_b = math.sqrt(sum(v**2 for v in vec_b.values()))

    if norm_a == 0 or norm_b == 0:
        return 0.0
    return dot_product / (norm_a * norm_b)

max_name_len = max(len(name) for name in doc_names)
label_width = max_name_len + 5

print("COSINE SIMILARITY ANTARDOKUMEN")
print("-" * (label_width + 35))

print(f"{'':<{label_width}}", end="")
for i in range(len(doc_names)):
    print(f" {'D'+str(i+1):>6}", end="")
print()
print("-" * (label_width + 35))


for i, name_i in enumerate(doc_names):
    label = f"D{i+1} {name_i}"
    print(f"{label:<{label_width}}", end="")
    for j in range(len(doc_names)):
        sim = calculate_cosine(tfidf_matrix[i], tfidf_matrix[j])
        print(f" {sim:>6.4f}", end="")
    print()

print()
print("Interpretasi (urut dari paling mirip):")
sims = []
for i in range(len(doc_names)):
    for j in range(i+1, len(doc_names)):
        sim = calculate_cosine(tfidf_matrix[i], tfidf_matrix[j])
        sims.append((sim, i, j))
sims.sort(key=lambda x: x[0], reverse=True)

for sim, i, j in sims:
    interpretasi = ("sangat mirip" if sim >= 0.5
                    else "cukup mirip" if sim >= 0.2
                    else "kurang mirip")
    print(f"  D{i+1} vs D{j+1}  ({doc_names[i]:<20} <-> {doc_names[j]:<20})  "
          f"sim={sim:.4f}  {interpretasi}")
```
